# 📖 Text Readability Analyzer

## 🎯 Objective

Estimate how easy or difficult a piece of English text is to read using the Flesch Reading Ease score.


```python
!pip install textstat
```

```python
import textstat

text = input("Enter an English paragraph: ")

if not text.strip():
    print("Please enter some text.")
else:
    score = textstat.flesch_reading_ease(text)
    grade = textstat.flesch_kincaid_grade(text)

    print("\n===== READABILITY ANALYZER =====")
    print(f"Flesch Reading Ease Score: {score:.2f}")
    print(f"Estimated Grade Level: {grade:.2f}")

    if score >= 90:
        level = "Very Easy"
    elif score >= 80:
        level = "Easy"
    elif score >= 70:
        level = "Fairly Easy"
    elif score >= 60:
        level = "Standard"
    elif score >= 50:
        level = "Fairly Difficult"
    elif score >= 30:
        level = "Difficult"
    else:
        level = "Very Difficult"

    print("Reading Difficulty:", level)
```

**Input:** `Python is a simple programming language. It is easy to learn and useful for many applications.`

The program returns a readability score and an estimated grade level.
